```python
import requests
import ollama

# Define a web search tool
def web_search(query: str):
    # This example uses DuckDuckGo's API; replace with your preferred search API
    response = requests.get(
        "https://api.duckduckgo.com/",
        params={"q": query, "format": "json"}
    )
    data = response.json()
    return data.get("AbstractText", "No results found.")

# Define the tool for Ollama
tools = [
    {
        "type": "function",
        "function": {
            "name": "web_search",
            "description": "Search the web for live information",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {"type": "string"}
                },
                "required": ["query"]
            }
        }
    }
]

# Example chat with tool calling
messages = [
    {"role": "user", "content": "What are the latest developments in AI ethics?"}
]

response = ollama.chat(
    model="gemma3:1b", # Or another model supporting tool calling
    messages=messages,
    tools=tools
)

print(response['message']['content'])
```
